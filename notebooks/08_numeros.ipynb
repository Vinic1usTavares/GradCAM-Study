## Resumo das Métricas Computadas

Exibindo os dados de discriminação de classe carregados do cache.

```python
metrics = deck.result("class_discrimination")
agg = metrics["aggregate"]
fig, ax = plt.subplots(figsize=(6, 4), dpi=110)
methods = list(agg.keys())
values = [agg[m]["mean"] for m in methods]
ax.bar(methods, values, color=['gray', 'blue', 'green'])
ax.set_ylim(0, 1.1)
ax.set_title("Discriminação de Classe (Spearman)")
deck.show(fig)
```
